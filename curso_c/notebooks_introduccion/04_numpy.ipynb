## 1. Vectores y Matrices (Arrays)
A diferencia de las listas tradicionales de Python, los arrays de NumPy son homogéneos (contienen un solo tipo de datos) y permiten operaciones matemáticas vectorizadas elemento a elemento.

```python
import numpy as np

# Crear un array a partir de una lista
temperaturas = np.array([15.2, 18.1, 22.0, 19.5, 14.8])
print('Temperaturas:', temperaturas)
print('Tipo de dato:', temperaturas.dtype)
print('Forma del array:', temperaturas.shape)
```

### Consigna 4.3: Filtrado de Lluvias con Máscaras Booleanas
Las máscaras booleanas permiten seleccionar rápidamente elementos de un array que cumplen con una condición lógica.
Dado el registro de lluvias en milímetros de un mes, filtra y obtén un nuevo array únicamente con los valores de precipitación mayores o iguales a 5 mm.

```python
import numpy as np
registro_lluvias = np.array([0.0, 12.4, 4.2, 0.0, 25.1, 1.5, 8.0, 0.0, 3.2, 15.6])

# TU CÓDIGO AQUÍ: Aplica la máscara condicional sobre el array
lluvias_significativas = registro_lluvias[registro_lluvias >= 5.0]

print('Días con lluvias significativas (>= 5 mm):', lluvias_significativas)
```

```python
# CELDA DE VERIFICACIÓN
try:
    assert len(lluvias_significativas) == 4, 'La cantidad de días con lluvias significativas no es correcta.'
    assert np.all(lluvias_significativas >= 5.0), 'El filtrado contiene valores menores a 5.0 mm.'
    print('¡Felicidades! Desafío 4.2 (Filtrado de datos con NumPy) superado. 🌧️')
except AssertionError as e:
    print(f'Error de validación: {e}')
```
